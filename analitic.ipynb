```python
import numpy as np
from scipy.integrate import solve_ivp
import plotly.graph_objects as go

# ------------------- Parámetros -------------------
m, e = 1.0, 1.0
q0, p0 = 0.0, 1.0
t_span = (0, 40)
t_eval = np.linspace(0, 40, 15000)      # muchos puntos = curva muy suave

# ------------------- Campo eléctrico -------------------
def hamilton_eqs(t, X):
    q, p = X
    # ← Aquí cambias el campo cuando quieras
import numpy as np
from scipy.integrate import solve_ivp
import plotly.graph_objects as go

# ------------------- Parámetros -------------------
m, e = 1.0, 1.0
q0, p0 = 0.0, 1.0
t_span = (0, 40)
t_eval = np.linspace(0, 40, 15000)      # muchos puntos = curva muy suave

# ------------------- Campo eléctrico -------------------
def hamilton_eqs(t, X):
    q, p = X
    # ← Aquí cambia    E = np.sin(0.5 * t)                  # el que estabas usando
    # E = 0.4                              # constante
    # E = 1.5 * np.cos(1.2 * t)            # casi resonante → precioso caos
    # E = 5.0 * np.exp(- (t-10)**2 / 2)     # pulso gaussiano
    return [p/m, e*E]

# ------------------- Integración -------------------
sol = solve_ivp(hamilton_eqs, t_span, [q0, p0],
                t_eval=t_eval, rtol=1e-10, atol=1e-12)

t = sol.t
q = sol.y[0]
p = sol.y[1]
v = p / m                                 # velocidad = p (m=1)

# ==================== 1. Trayectoria 3D (q, p, t) ====================
fig1 = go.Figure()

fig1.add_trace(go.Scatter3d(
    x=q, y=p, z=t,
    mode='lines',
    line=dict(
        color=t,                    
        colorscale='Inferno',       # elige: Viridis, Plasma, Inferno, Rainbow, Jet...
        width=8,
        colorbar=dict(title="tiempo t", thickness=15)
    ),
    name='trayectoria',
    hovertemplate='<b>q</b>: %{x:.3f}<br><b>p</b>: %{y:.3f}<br><b>t</b>: %{z:.2f}<extra></extra>'
))

fig1.update_layout(
    title=dict(text="Trayectoria 3D interactiva (q, p, t) — rota y haz zoom libremente",
               x=0.5, font=dict(size=18)),
    scene=dict(
        xaxis_title='posición q',
        yaxis_title='momento p',
        zaxis_title='tiempo t',
        aspectmode='manual',
        aspectratio=dict(x=1, y=1, z=0.6)
    ),
    width=1000,
    height=800
)

fig1.show()

# ==================== 2. Otra vista 3D (q, v, t) ====================
fig2 = go.Figure()

fig2.add_trace(go.Scatter3d(
    x=q, y=v, z=t,
    mode='lines',
    line=dict(
        color=t,
        colorscale='Inferno',
        width=8,
        colorbar=dict(title="tiempo t")
    ),
    hovertemplate='q: %{x:.3f}<br>v: %{y:.3f}<br>t: %{z:.2f}<extra></extra>'
))

fig2.update_layout(
    title="Trayectoria 3D (q, velocidad v, t)",
    scene=dict(xaxis_title='q', yaxis_title='velocidad v', zaxis_title='tiempo t'),
    width=1000, height=800
)

fig2.show()

# ==================== 3. Espacio de fases 2D coloreado por tiempo ====================
fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=q, y=p,
    mode='markers',
    marker=dict(
        color=t,
        colorscale='Inferno',
        size=4,
        colorbar=dict(title="tiempo t")
    ),
    hovertemplate='t = %{marker.color:.2f}<extra></extra>'
))

# Punto inicial destacado
fig3.add_trace(go.Scatter(
    x=[q[0]], y=[p[0]],
    mode='markers',
    marker=dict(size=12, color='red', symbol='circle'),
    name='inicio (t=0)'
))

fig3.update_layout(
    title="Espacio de fases (q,p) coloreado por tiempo",
    xaxis_title="q", yaxis_title="p",
    width=900, height=700,
    showlegend=False
)
fig3.show()

# ==================== Guardar como HTML interactivo (opcional) ====================
fig1.write_html("trayectoria_3D_qpt.html")
fig2.write_html("trayectoria_3D_qvt.html")
fig3.write_html("fases_qp.html")
```

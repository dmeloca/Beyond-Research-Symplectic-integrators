# Unified Graph

```python
from tools import make_field, simulate_strang, simulate_rk4, simulate_analytic
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8")
```

```python
E, E_dot = make_field(omega=0.5)

q0, p0, t0, E0 = 0.0, 1.0, 0.0, 0.0
dt = 0.01
steps = 4000

t_span = (t0, t0 + steps * dt)
t_eval = np.linspace(t_span[0], t_span[1], steps + 1)


traj_s, mom_s, times_s, H_s, Hext_s = simulate_strang(
    q0, p0, t0, E0, dt, steps, E, E_dot
)
traj_r, mom_r, times_r, H_r = simulate_rk4(
    q0, p0, t0, dt, steps, E
)
traj_a, mom_a, times_a, H_a = simulate_analytic(
    q0, p0, t_span, t_eval, E
)
```

```python
plt.figure(figsize=(6,6))
plt.plot(traj_s, mom_s, 'r-', label="Strang")     
plt.plot(traj_r, mom_r, 'g--', label="RK4")
plt.plot(traj_a, mom_a, 'b--', label="Analytic")

plt.title("Espacio de fase (q, p)")
plt.xlabel("q (posición)")
plt.ylabel("p (momento)")

plt.legend()
plt.grid(True)
#plt.axis("equal")   
plt.show()
```
